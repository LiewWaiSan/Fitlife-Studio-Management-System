Task 1 · Group Members by Plan + Count Each Band

1. Create three counter variables, all starting at 0: basic_count, plus_count, premium_count.
2. Use a for-loop to walk through every member dictionary in the members list.
3. Inside the loop, use an if / elif / elif / else chain on the member's plan to:
4. a) increment the matching counter,
5. b) print one line per member in this exact form: FL-001 Tan Jia Hui [Basic]
6. After the loop, print a Class Stats card with the three c

```python
# ==========================================
# FITLIFE PROJECT - MILESTONE 3
# MEMBER DATA FROM MILESTONE 2
# ==========================================

members = [
    {
        "id": "FL-001",
        "name": "Tan Jia Hui",
        "plan": "Basic",
        "expires": "2026-04-15",
        "status": "EXPIRED"
    },
    {
        "id": "FL-002",
        "name": "Rajesh Kumaran",
        "plan": "Premium",
        "expires": "2026-08-03",
        "status": "active"
    },
    {
        "id": "FL-003",
        "name": "Nur Hidayah Binte Rahman",
        "plan": "Plus",
        "expires": "2026-05-10",
        "status": "EXPIRED"
    },
    {
        "id": "FL-004",
        "name": "Marcus Wong",
        "plan": "Basic",
        "expires": "2026-10-05",
        "status": "active"
    },
    {
        "id": "FL-005",
        "name": "Priya Subramaniam",
        "plan": "Premium",
        "expires": "2026-07-01",
        "status": "active"
    },
    {
        "id": "FL-006",
        "name": "David Chen",
        "plan": "Plus",
        "expires": "2026-10-20",
        "status": "active"
    }
]

print("FitLife member data loaded successfully.")
print("Total members:", len(members))
```

```python
# ==========================================
# TASK 1 - GROUP MEMBERS BY PLAN
# ==========================================

basic_count = 0
plus_count = 0
premium_count = 0

for member in members:

    if member["plan"] == "Basic":
        basic_count += 1

    elif member["plan"] == "Plus":
        plus_count += 1

    elif member["plan"] == "Premium":
        premium_count += 1

    else:
        print("Unknown plan")

    print(f"{member['id']} {member['name']:<24} [{member['plan']}]")

print()
print("============= MEMBER STATS =============")
print(f" Basic   : {basic_count}")
print(f" Plus    : {plus_count}")
print(f" Premium : {premium_count}")
print(f" Total   : {len(members)}")
print("========================================")
```

Task 2 · Class Booking Eligibility (and / or)
1. Use a for-loop to walk through every member.
2. Inside the loop, build a boolean variable plan_allows_class = (member['plan'] != 'Basic'). (Plus and
Premium can book; Basic cannot.)
3. Build a second boolean: is_active = (member['status'] == 'active').
4. Combine them with AND: can_book = plan_allows_class and is_active.
5. Use if / else to print one line per member:
6. if can_book: print(f" {name:<24} - can book")
7. else: print(f" {name:<24} - CANNOT book")
8. After the loop, print a one-line summary of how many CAN book (use a counter from earlier inside the
loop).


```python
# ==========================================
# TASK 2 - CLASS BOOKING ELIGIBILITY
# ==========================================

can_book_count = 0

for member in members:

    plan_allows_class = member["plan"] != "Basic"
    is_active = member["status"] == "active"

    can_book = plan_allows_class and is_active

    if can_book:
        print(f" {member['name']:<24} - can book")
        can_book_count += 1
    else:
        print(f" {member['name']:<24} - CANNOT book")

print()
print(f"Total members who can book a class: {can_book_count}")
```

Task 3 · Renewal Loop with String Comparison
1. Use a for-loop to walk every member.
FitLife Project · Milestone 3 · Business Logic · PDDS-PYD ·
2. Inside the loop, compare member['expires'] with the string "2026-05-25" (today). The trick: string
comparison works alphabetically, so '2026-04-15' < '2026-05-25' is TRUE.
3. If the member is expired (expires < today):
4. a) Print Renewing FL-001 Tan Jia Hui (was 2026-04-15)
5. b) Compute the new expiry by adding 6 months. Use string slicing: take the year and month, increment
month by 6, watch out for month > 12. For the project, just compute year, month, day separately as ints,
then assemble back into a string.
6. c) Update member['expires'] in place.
7. d) Also set member['status'] = 'active' (the renewal makes them active again).
8. e) Print ... renewed to 2026-10-15
9. If the member is NOT expired, use continue to skip them.
10. After the loop, print a summary of how many were renewed.

```python
# ==========================================
# TASK 3 - RENEW EXPIRED MEMBERS
# ==========================================

today = "2026-05-25"
renewed_count = 0

for member in members:

    if member["expires"] >= today:
        continue

    old_expiry = member["expires"]

    print(
        f"Renewing {member['id']} "
        f"{member['name']} (was {old_expiry})"
    )

    # Extract year, month and day
    year = int(member["expires"][:4])
    month = int(member["expires"][5:7])
    day = int(member["expires"][8:10])

    # Add 6 months
    new_month = month + 6

    # Handle next year
    if new_month > 12:
        new_month -= 12
        year += 1

    # Create new expiry date
    new_expiry = f"{year:04d}-{new_month:02d}-{day:02d}"

    # Update member
    member["expires"] = new_expiry
    member["status"] = "active"

    renewed_count += 1

    print(f" ... renewed to {new_expiry}")

print()
print(f"Total members renewed today: {renewed_count}")
```

Task 4 · Build the Expired-Members List
1. Create an empty list: expired_list = [].
2. Use a for-loop to walk the members.
3. Inside the loop, if member['expires'] < '2026-05-25', append the member's name AND expires date to
expired_list (as a tuple or a small string - your choice).
4. After the loop, print a section header ===== EXPIRED MEMBERS ===== and then one line per expired
member in the form: - Tan Jia Hui (expired 2026-04-15).
5. Print one final line: Total expired: 2.


```python
# ==========================================
# TASK 4 - EXPIRED MEMBERS LIST
# ==========================================

expired_list = []

for member in members:

    if member["expires"] < "2026-05-25":

        expired_list.append(
            (member["name"], member["expires"])
        )

print("===== EXPIRED MEMBERS =====")

for expired_member in expired_list:

    name = expired_member[0]
    expires = expired_member[1]

    print(f" - {name} (expired {expires})")

print(f"Total expired: {len(expired_list)}")
```

Task 5 · Total Monthly Fees (Counter + if/elif inside a loop)
1. Create a variable total_fees = 0 (an integer for now).
2. Use a for-loop to walk every member.
3. Inside the loop, use an if / elif / else chain on the plan:
4. if plan == "Basic": total_fees += 88
5. elif plan == "Plus": total_fees += 128
6. elif plan == "Premium": total_fees += 188
7. After the loop, print a one-line summary in this exact form: Total monthly fees: S$808.00.



```python
# ==========================================
# TASK 5 - TOTAL MONTHLY FEES
# ==========================================

total_fees = 0

for member in members:

    if member["plan"] == "Basic":
        total_fees += 88

    elif member["plan"] == "Plus":
        total_fees += 128

    elif member["plan"] == "Premium":
        total_fees += 188

    else:
        print("Unknown plan")

print(f"Total monthly fees: S${total_fees:,.2f}")
print()
print(
    "(2 x Basic + 2 x Plus + 2 x Premium "
    "= 176 + 256 + 376 = S$808.00)"
)
```
