Task 1 · Build enrol_member()
1. At the TOP of the cell, import datetime:  from datetime import date, timedelta.
2. Define a function called enrol_member with this exact signature:  enrol_member(name, plan="Basic",
branch="Tampines One", months=6).
3. Inside the function:
4.     a) Generate a NEW member ID. Use the next number after FL-006 (so the next is FL-007, then FL-008,
etc.). Use a global counter variable next_id_counter that starts at 7 BEFORE the function definition.
5.     b) Compute the start_date as date.today() and the expires_date as start_date + timedelta(days=30 *
months).
6.     c) Build a dictionary with all the standard keys (id, name, plan, branch, joined, expires, status="active").
7.     d) Print a one-line confirmation:  Enrolled FL-007 Aisha Tan on the Basic plan at Tampines One. Expires:
2026-12-12.
8.     e) Return the new member dictionary.
9. Call enrol_member three times to enrol the three walk-ins from the sample data, each one
demonstrating a different mix of defaults vs explicit args.

```python
from datetime import date, timedelta

next_id_counter = 7

def enrol_member(name, plan="Basic", branch="Tampines One", months=6):

    global next_id_counter

    # Generate new ID
    member_id = f"FL-{next_id_counter:03d}"
    next_id_counter += 1

    # Calculate dates
    start_date = date.today()
    expires_date = start_date + timedelta(days=30 * months)

    # Create member dictionary
    new_member = {
        "id": member_id,
        "name": name,
        "plan": plan,
        "branch": branch,
        "joined": start_date.isoformat(),
        "expires": expires_date.isoformat(),
        "status": "active"
    }

    print(
        f"Enrolled {member_id} {name} on the {plan} plan "
        f"at {branch}. Expires: {expires_date.isoformat()}."
    )

    return new_member


# Enrol the three new members
m7 = enrol_member("Aisha Tan")

m8 = enrol_member(
    "Chong Wei Han",
    plan="Plus"
)

m9 = enrol_member(
    "Ravi Pillai",
    plan="Premium",
    branch="JEM"
)
```

Task 2 — renew_membership()
1. Function signature:  renew_membership(member, months=6).
2. Inside the function:
3.     a) Parse the current expires string into a date object:  date.fromisoformat(member['expires']).
4.     b) Compute the new expiry:  new_expiry = old_expiry + timedelta(days=30 * months).
5.     c) Convert back to ISO string:  member['expires'] = new_expiry.isoformat().
FitLife Project · Milestone 4 · Reusable Function Library · PDDS-PYD ·  
6.     d) Set member['status'] = 'active'.
7.     e) Print a confirmation:  Renewed FL-001 Tan Jia Hui. New expiry: 2026-10-15. Status: active.
8.     f) Return the updated member.
9. Call renew_membership on the two expired members from M3 (FL-001 and FL-003).

```python
from datetime import date, timedelta

def renew_membership(member, months=6):

    # Convert expiry string into date
    old_expiry = date.fromisoformat(member["expires"])

    # Add renewal period
    new_expiry = old_expiry + timedelta(days=30 * months)

    # Update member
    member["expires"] = new_expiry.isoformat()
    member["status"] = "active"

    print(
        f"Renewed {member['id']} {member['name']}. "
        f"New expiry: {member['expires']}. "
        f"Status: {member['status']}."
    )

    return member
```

```python
member1 = {
    "id": "FL-001",
    "name": "Tan Jia Hui",
    "plan": "Basic",
    "expires": "2026-04-15",
    "status": "EXPIRED"
}

member3 = {
    "id": "FL-003",
    "name": "Nur Hidayah Binte Rahman",
    "plan": "Plus",
    "expires": "2026-05-10",
    "status": "EXPIRED"
}

renew_membership(member1)
renew_membership(member3)
```

Task 3 · Build calculate_fees() that returns multiple values
1. Function signature:  calculate_fees(members_list).
2. Inside the function, define a small lookup dictionary  fee_table = {"Basic": 88, "Plus": 128, "Premium":
188}.
3. Use a for-loop to walk every member, look up their fee using fee_table[member['plan']], and accumulate
four running totals: grand_total, basic_total, plus_total, premium_total.
4. Return all four values in one return:  return grand_total, basic_total, plus_total, premium_total.
5. Call calculate_fees and UNPACK the four returned values into four named variables on a single line.
6. Print a Fee Summary card with all four totals, each in S$X,XXX.00 format.

```python
members = [
    {
        "id": "FL-001",
        "name": "Tan Jia Hui",
        "plan": "Basic",
        "branch": "Tampines One",
        "joined": "2025-10-15",
        "expires": "2026-04-15",
        "status": "EXPIRED"
    },

    {
        "id": "FL-002",
        "name": "Rajesh Kumaran",
        "plan": "Premium",
        "branch": "Bukit Timah Plaza",
        "joined": "2026-02-03",
        "expires": "2026-08-03",
        "status": "active"
    },

    {
        "id": "FL-003",
        "name": "Nur Hidayah Binte Rahman",
        "plan": "Plus",
        "branch": "Tampines One",
        "joined": "2025-11-10",
        "expires": "2026-05-10",
        "status": "EXPIRED"
    },

    {
        "id": "FL-004",
        "name": "Marcus Wong",
        "plan": "Basic",
        "branch": "JEM",
        "joined": "2026-04-05",
        "expires": "2026-10-05",
        "status": "active"
    },

    {
        "id": "FL-005",
        "name": "Priya Subramaniam",
        "plan": "Premium",
        "branch": "Tampines One",
        "joined": "2026-01-01",
        "expires": "2026-07-01",
        "status": "active"
    },

    {
        "id": "FL-006",
        "name": "David Chen",
        "plan": "Plus",
        "branch": "Bukit Timah Plaza",
        "joined": "2026-04-20",
        "expires": "2026-10-20",
        "status": "active"
    }
]
```

```python
def calculate_fees(members_list):

    fee_table = {
        "Basic": 88,
        "Plus": 128,
        "Premium": 188
    }

    grand_total = 0
    basic_total = 0
    plus_total = 0
    premium_total = 0

    for member in members_list:

        fee = fee_table[member["plan"]]

        grand_total += fee

        if member["plan"] == "Basic":
            basic_total += fee

        elif member["plan"] == "Plus":
            plus_total += fee

        elif member["plan"] == "Premium":
            premium_total += fee

    return grand_total, basic_total, plus_total, premium_total
```

```python
total, basic, plus, premium = calculate_fees(members)

print("=========== FEE SUMMARY ===========")
print(f" Grand total : S${total:,.2f}")
print(f" Basic       : S${basic:,.2f}")
print(f" Plus        : S${plus:,.2f}")
print(f" Premium     : S${premium:,.2f}")
print("===================================")
```

Task 4 · Build check_eligibility() that returns True/False
1. Function signature:  check_eligibility(member).
2. Inside the function, use the SAME AND logic from M3 Task 2:
3.     plan_allows = member['plan'] != 'Basic'
4.     is_active   = member['status'] == 'active'
5.     return plan_allows and is_active
6. Call check_eligibility on EACH of the six original members. Use a for-loop, but the inside of the loop just
calls the function and prints the result. The function does the heavy lifting.

```python
def check_eligibility(member):

    plan_allows = member["plan"] != "Basic"

    is_active = member["status"] == "active"

    return plan_allows and is_active
```

```python
for member in members:

    result = check_eligibility(member)

    print(
        f"{member['name']:<25} - eligible: {result}"
    )
```

Task 6 · Tie It All Together + Save the Final Report to a File

```python
from datetime import datetime, date, timedelta

# Calculate fees
total, basic, plus, premium = calculate_fees(members)

# WRITE REPORT
with open("members_report.txt", "w") as f:

    f.write("============= FITLIFE CLASS REPORT =============\n")

    f.write(
        "Generated: "
        + datetime.now().strftime("%Y-%m-%d %H:%M")
        + "\n"
    )

    f.write(f"Total members: {len(members)}\n")
    f.write("\n")

    # Member details
    for member in members:

        f.write(
            f"{member['id']} | "
            f"{member['name']} | "
            f"{member['plan']} | "
            f"{member['status']}\n"
        )

    # Fee summary
    f.write("\nFEES\n")

    f.write(f" Grand total : S${total:,.2f}\n")
    f.write(f" Basic       : S${basic:,.2f}\n")
    f.write(f" Plus        : S${plus:,.2f}\n")
    f.write(f" Premium     : S${premium:,.2f}\n")

    # Eligible members
    f.write("\nELIGIBLE FOR CLASS BOOKING\n")

    for member in members:

        if check_eligibility(member):

            f.write(
                f" {member['name']}\n"
            )

    f.write("\n----- end of report -----\n")


# READ REPORT
with open("members_report.txt", "r") as f:

    print(f.read())
```
